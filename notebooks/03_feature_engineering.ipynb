## 2. Feature Correlation with Target Yield/Production

```python
features_to_check = [
    'harvested_000_acres', 'sown_000_acres', 'yield_lag1', 'yield_lag2',
    'rainfall_mm', 'rainfall_roll3', 'temp_roll3', 'rain_x_temp', 'inf_x_gdp',
    'drought_flag', 'harvest_efficiency', 'season_binary', 'temperature_c',
    'gdp_billion_usd', 'inflation_pct'
]

target_corrs = df_feat[features_to_check + ['target_yield_production']].corr()['target_yield_production'].drop('target_yield_production').sort_values(ascending=True)

plt.figure(figsize=(10, 8))
target_corrs.plot(kind='barh', color='#3498db', edgecolor='black')
plt.title('Feature Correlations with Seasonal Target Production', fontsize=14, fontweight='bold')
plt.xlabel('Pearson Correlation Coefficient (r)')
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```
